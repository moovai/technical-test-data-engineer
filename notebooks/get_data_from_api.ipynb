```python
import pandas as pd  
import requests 
```

```python
BASE_URL = "http://127.0.0.1:8001"  
endpoint_users= "/users"
endpoint_tracks = "/tracks"
endpoint_history = "/listen_history"
```

```python
response = requests.get(f"{BASE_URL}{endpoint_users}")
if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame(data)
    print(df.head())
else:
    print(f"Error: {response.status_code} - {response.text}")
```

```python
response = requests.get(f"{BASE_URL}{endpoint_tracks}")
if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame(data)
    print(df.head())
else:
    print(f"Error: {response.status_code} - {response.text}")
```

```python
response = requests.get(f"{BASE_URL}{endpoint_history}")
if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame(data)
    print(df.head())
else:
    print(f"Error: {response.status_code} - {response.text}")
```
